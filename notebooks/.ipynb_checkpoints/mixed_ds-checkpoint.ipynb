```python
import polars as pl
import numpy as np
import polars.selectors as sl
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from sentence_transformers import SentenceTransformer, util 
from itertools import combinations
import re

pl.Config.set_fmt_str_lengths(500)
pl.Config.set_tbl_rows(-1)
```

```python
feret_path = Path('../data/feret/OFIQ')
feret_list = []

for f in sorted(feret_path.iterdir()):
    if f.is_file() and f.suffix == '.csv':
        temp = pl.read_csv(f, separator=';')
        feret_list.append(temp)

feret = pl.concat(feret_list, how='vertical_relaxed')
feret.head(5)
```

```python
cfd_path = Path('../data/cfd/OFIQ')
cfd_list = []

for f in sorted(cfd_path.iterdir()):
    if f.is_file() and f.suffix == '.csv':
        temp = pl.read_csv(f, separator=';', infer_schema_length=None,)
        cfd_list.append(temp)

cfd = pl.concat(cfd_list, how='vertical_relaxed')
cfd.head(5)
```

### Subject related columns

```python
subject_columns = [
    "HeadPoseYawFrontal", 
    "HeadPosePitchFrontal", 
    "HeadPoseRollFrontal", 
    "EyesOpen", 
    "EyesVisible", 
    "ExpressionNeutrality", 
    "MouthClosed", 
    "MouthOcclusionPrevention", 
    "FaceOcclusionPrevention", 
    "NoHeadCoverings", 
    "SingleFacePresent"
 ]

framing_columns = [
    "HeadSize", 
    "InterEyeDistance", 
    "LeftwardCropOfFace", 
    "RightwardCropOfFace", 
    "MarginAboveTheFace", 
    "MarginBelowTheFace"
]

scene_columns = [
    "UnifiedQualityScore", 
    "BackgroundUniformity", 
    "IlluminationUniformity", 
    "LuminanceMean", 
    "LuminanceVariance", 
    "UnderExposurePrevention", 
    "OverExposurePrevention", 
    "DynamicRange", 
    "Sharpness", 
    "NoCompressionArtifacts", 
    "NaturalColour"
]

all_subject_columns = [f"{n}.scalar" for n in subject_columns]
```

```python
subject_feret = feret.select(pl.col(['Filename',*all_subject_columns]))
subject_feret.head(3) 
```

```python
feret_recordings_path = Path('../data/feret/dvd1_recordings.csv')

feret_recordings = pl.read_csv(
    feret_recordings_path,
    separator = ',',
    infer_schema_length=None
)

feret_recordings.head(5)
```

```python
rec_col = [c for c in feret_recordings.columns if re.search(r"yaw|pitch|roll|glass|beard|must|expr|pose|file|name", c, re.I)]
rec_col
```

```python
sub_fer_rec = feret_recordings.select(pl.col(['URL_relative',*rec_col]))
sub_fer_rec.head(3)
```

```python
rec_sel = feret_recordings.select(
    pl.col("URL_relative").str.split("/").list.last().str.replace(r"\.ppm(\.bz2)?$", "").alias("stem"), 
    pl.col("Subject_id"), 
    pl.col("Subject_Application_Face_Pose_name").alias("pose_gt"), 
    pl.col("Subject_Application_Face_Pose_yaw").alias("yaw_nom"), 
    pl.col("Subject_Application_Face_Pose_pitch").alias("pitch_nom"), 
    pl.col("Subject_Application_Face_Pose_roll").alias("roll_nom"), 
    pl.col("Subject_Application_Face_Wearing_glasses").alias("glasses"), 
    pl.col("Subject_Application_Face_Hair_beard").alias("beard"), 
    pl.col("Subject_Application_Face_Hair_mustache").alias("mustache"),
    pl.col("Subject_Application_Face_Expression_name").alias("expression")
)

ofiq_k = subject_feret.with_columns(
    pl.col("Filename").str.split("/").list.last().str.replace(r"\.(png|ppm)$", "").alias("stem")
)
j = ofiq_k.join(rec_sel, on="stem", how="left")
```

```python
j.head(10)
```

```python
rows = [f"{n}.scalar" for n in subject_columns]

d = j.with_columns(
    [pl.when(pl.col(c) >= 0).then(pl.col(c)).alias(c) for c in rows]
).with_columns(
    (pl.col("glasses") == "Yes").cast(pl.Int8).alias("glasses_yes"),
    (pl.col("beard") == "Yes").cast(pl.Int8).alias("beard_yes"), 
    (pl.col("mustache") == "Yes").cast(pl.Int8).alias("mustache_yes"),
    pl.col("yaw_nom").abs().alias("abs_nominal_yaw"), 
    (pl.col("expression") == "fb").cast(pl.Int8).alias("alt_expression")
)

def corr_matrix(d, rows, cols, method="spearman"):
    out = []
    for r in rows:
        row = {"component": r.replace(".scalar", "")}
        for c in cols:
            sub = d.select(pl.col(r), pl.col(c)).drop_nulls()
            row[c] = sub.select(pl.corr(r, c, method=method)).item()
        out.append(row)
    return pl.DataFrame(out)

def plot_heatmap(M, title):
    cols = [c for c in M.columns if c != "component"]
    A = M.select(cols).fill_null(np.nan).to_numpy()
    fig, ax = plt.subplots(figsize=(1.4 * len(cols) + 4, 0.45 * M.height + 2))
    sns.heatmap(A, annot=True, fmt=".2f", cmap="coolwarm", center=0, vmin=-1, vmax=1, xticklabels=cols, yticklabels=M["component"].to_list(), ax=ax)
    ax.set_title(title)
    plt.tight_layout()
    plt.show()

plot_heatmap(corr_matrix(d, rows, ["glasses_yes", "beard_yes", "mustache_yes", "abs_nominal_yaw"]), "OFIQ scalar vs FERET labels (Spearman), all poses")

frontal = d.filter(pl.col("pose_gt").is_in(["fa", "fb"]))
plot_heatmap(
    corr_matrix(
        frontal, 
        rows, 
        ["glasses_yes", "beard_yes", "mustache_yes", "alt_expression"])
    , "OFIQ scalar vs FERET labels (Spearman), frontal poses only (fa/fb)");
```

```python
fr = d.filter(pl.col("pose_gt").is_in(["fa", "fb"]))

def roc_table(df, comp, label, thresholds):
    pos = df.filter(pl.col(label) == 1)[comp].drop_nulls()
    neg = df.filter(pl.col(label) == 0)[comp].drop_nulls()
    print("positives:", pos.len(), "negatives:", neg.len())
    rows = []
    for t in thresholds:
        tpr, fpr = (pos < t).mean(), (neg < t).mean()
        rows.append({"threshold": t, "TPR": tpr, "FPR": fpr, "J": tpr - fpr})
    return pl.DataFrame(rows)

print(roc_table(fr, "EyesVisible.scalar", "glasses_yes", [50, 60, 70, 80, 90, 95, 99, 100]))
```

```python
print(d.group_by("pose_gt").agg(pl.len().alias("n"), 
                                pl.col("HeadPoseYawFrontal.scalar").median().alias("yaw_scalar_median"), 
                                pl.col("HeadPoseYawFrontal.scalar").quantile(0.05).alias("p5"), 
                                pl.col("HeadPoseYawFrontal.scalar").quantile(0.95).alias("p95"), 
                                (pl.col("HeadPoseYawFrontal.scalar") <= 1)
                                    .mean().alias("share_scalar_0")).sort("yaw_scalar_median", descending=True))
```
